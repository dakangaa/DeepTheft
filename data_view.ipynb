```python
import h5py
from collections import defaultdict
import matplotlib.pyplot as plt
# 查看traces长度分布
file = h5py.File(r"dataset/linear.h5", "r")
count = defaultdict(int)
traces = file["trace"]
for k in traces.keys():
    count[traces[k].shape[0]] += 1
plt.bar(count.keys(), count.values())
```

```python
print(max(count, key=lambda k : count[k]))
print(min(count))
print(max(count))
```

```python
hp = file["hp"]
hp_distrib = defaultdict(int)
for k in traces.keys():
    hp_distrib[hp[k][1]] += 1
plt.bar(hp_distrib.keys(), hp_distrib.values())
```

```python
# 查看K，S的分布
file = h5py.File(r"dataset/max_pool2d.h5", "r")
count_K = defaultdict(int)
count_S = defaultdict(int)
count_P = defaultdict(int)
hps = file["hp"]
for k in hps.keys():
    count_K[hps[k][2]] += 1
    count_S[hps[k][3]] += 1
    count_P[hps[k][4]] += 1
```

```python
print(count_K.keys())
plt.bar(count_K.keys(), count_K.values())
plt.gca().xaxis.set_major_locator(plt.MaxNLocator(integer=True))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel("results/parameter_sensitivity.xlsx", sheet_name="p")
df = df.set_index("test_domain")
f1_p = df.loc["mean"].to_numpy()
df = pd.read_excel("results/parameter_sensitivity.xlsx", sheet_name="w")
df = df.set_index("test_domain")
f1_w = df.loc["mean"].to_numpy()
df = pd.read_excel("results/parameter_sensitivity.xlsx", sheet_name="t")
df = df.set_index("test_domain")
f1_t = df.loc["mean"].to_numpy()

ws = [0.1, 0.5, 1.0, 2.0, 5.0]
ps = [0.5, 0.8, 0.9, 0.95, 0.99]
ts = [0.01, 0.05, 0.1, 0.2, 0.5]

# 封装一个画图 + 标注函数（更干净）
def plot_with_labels(x_values, y_values, xlabel):
    fig, ax = plt.subplots(figsize=(8, 2))
    x_pos = range(len(x_values))

    ax.plot(x_pos, y_values)
    ax.set_ylim(95, 100)
    ax.set_xlabel(xlabel)

    ax.set_xticks(x_pos)
    ax.set_xticklabels([str(x) for x in x_values])
    # 给每个点加数值
    for xi, yi in zip(x_pos, y_values):
        ax.annotate(f"{yi:.2f}", (xi, yi), textcoords="offset points",
                    xytext=(0, -10), ha='center', fontsize=8)
    return fig

# 三个子图
fig1 = plot_with_labels( ps, f1_p, r"$\alpha$")
fig1.savefig("results/parameter_sensitivity_p.png", bbox_inches='tight', dpi=400)
fig2 = plot_with_labels( ws, f1_w, r"$\lambda$")
fig2.savefig("results/parameter_sensitivity_w.png", bbox_inches='tight', dpi=400)
fig3 = plot_with_labels( ts, f1_t, r"$\tau$")
fig3.savefig("results/parameter_sensitivity_t.png", bbox_inches='tight', dpi=400)

plt.show()
```
