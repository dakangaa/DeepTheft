```python
import h5py
from collections import defaultdict
import matplotlib.pyplot as plt
# 查看traces长度分布
file = h5py.File(r"dataset/linear.h5", "r")
count = defaultdict(int)
traces = file["trace"]
for k in traces.keys():
    count[traces[k].shape[0]] += 1
plt.bar(count.keys(), count.values())
```

```python
print(max(count, key=lambda k : count[k]))
print(min(count))
print(max(count))
```

```python
hp = file["hp"]
hp_distrib = defaultdict(int)
for k in traces.keys():
    hp_distrib[hp[k][1]] += 1
plt.bar(hp_distrib.keys(), hp_distrib.values())
```

```python
print(hp_distrib)
```

```python
# 查看K，S的分布
file = h5py.File(r"dataset/max_pool2d.h5", "r")
count_K = defaultdict(int)
count_S = defaultdict(int)
count_P = defaultdict(int)
hps = file["hp"]
for k in hps.keys():
    count_K[hps[k][2]] += 1
    count_S[hps[k][3]] += 1
    count_P[hps[k][4]] += 1
```

```python
print(count_K.keys())
plt.bar(count_K.keys(), count_K.values())
plt.gca().xaxis.set_major_locator(plt.MaxNLocator(integer=True))
```

```python
print(count_S.keys())
plt.bar(count_S.keys(), count_S.values())
plt.gca().xaxis.set_major_locator(plt.MaxNLocator(integer=True))
```

```python
print(count_P.keys())
plt.bar(count_P.keys(), count_P.values())
plt.gca().xaxis.set_major_locator(plt.MaxNLocator(integer=True))
```
